#### DataFrame 
##### What is a DataFrame? 
##### A DataFrame is a two-dimensional table containing rows and columns. 
##### This is the structure you will use most often.

```python
data = {
    "Name" : ["one","two","three"],
    "Age" :[12,34,56,],
    "Marks" :[12,45,75]
}
df = pd.DataFrame(data)
print(df)
```
