```python
# Creating a New Column
# What does it do? 
# Creates a derived feature
df["passed"] = df["customer_age"] >= 30
```

```python
df["passed"]
```

```python
df["Bonus_Marks"] = df["customer_age"] * 0.10
#This creates a new feature based on an existing feature.
```

```python
df["Bonus_Marks"]
```

```python
df["Gender_code"] = df["gender"].map({
    "Male":1,
    "Female":0,
    "Non-Binary":3
})
```

```python
df["Gender_code"]
```

```python
# astype()
# Changes data type.
# Useful when a column has an incorrect or unsuitable type.
df["net_sales"] = df["net_sales"].astype(int)
```

```python
df["net_sales"]
```

```python
# pd.to_datetime() 
# Converts a column to datetime.
# This is important when working with: (1) Sales data ,(2)Transactions ,(3)Trips, (4)Time, (5)series, (6)Logs
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df["order_date"]
```

```python
df["Year"] = df["order_date"].dt.year
df["Month"] = df["order_date"].dt.month
df["Day"] = df["order_date"].dt.day
df["DayOfWeek"] = df["order_date"].dt.dayofweek
#This is another example of feature engineering.
# Each Df gives differnt values like if we call year it will show us years so we dont see unnecessary data
# if we call df month it will only show us months data 
```
