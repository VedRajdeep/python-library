```python
data = data[
    (data["Age"] >= 10) & (data["Age"] <= 20) &
    (data["Study_Hours"] >= 0)  & (data["Study_Hours"] <= 24) &
    (data["Attendance"] >= 0)  & (data["Attendance"] <= 100) &
    (data["Marks"] >= 0) & (data["Marks"] <= 100)
]
```

```python
data
```

```python
data[data["Marks"] > 80]
```

```python
data[data["Attendance"] > 75]
```

```python
data[data["Study_Hours"] > 5]
```

```python
Average_marks = data["Marks"].mean()
Median_marks = data["Marks"].median()
Average_Study_Hours = data["Study_Hours"].mean()
Highest_Marks = data["Marks"].max()
Lowest_Marks = data["Marks"].min()

print(f"Average marks: {Average_marks}")
print(f"Median marks: {Median_marks}")
print(f"Average study hours: {Average_Study_Hours}")
print(f"Highest marks: {Highest_Marks}")
print(f"Lowest marks: {Lowest_Marks}")
```

```python
data.groupby("Gender")["Marks"].mean()
```

```python
bins = [0, 80, 90, 100]
labels = ['Below 80%', '80%-90%', 'Above 90%']
data['Attendance_Group'] = pd.cut(data['Attendance'], bins=bins, labels=labels, include_lowest=True)
avg_by_attendance = data.groupby('Attendance_Group', observed=False)['Marks'].mean()
print(avg_by_attendance)
```

```python
marks_array = data['Marks'].to_numpy()
print(marks_array)
```

```python
mean_value = np.mean(marks_array)
print(mean_value)
```

```python
median_value = np.median(marks_array)
print(median_value)
```

```python
std_value = np.std(marks_array)
print(std_value)
```

```python
min_value = np.min(marks_array)
max_value = np.max(marks_array)
print(min_value)
print(max_value)
```

```python
curved_marks = np.clip(marks_array + 5, 0, 100)
print(curved_marks)
#The numpy.clip() function is used to restrict the values of an array within a specified minimum and maximum range
```

```python
filtered_marks = marks_array[marks_array > 80]
print(filtered_marks)
```

```python
small_array = np.array([1, 2, 3, 4, 5, 6])
reshaped_array = small_array.reshape(2, 3)
print(reshaped_array)
```

```python
arange_array = np.arange(0, 10, 2)
print(arange_array)
#np.arange(): This is a NumPy function used to generate sequences of numbers with a 
#fixed step size (similar to Python's built-in range() function, but optimized for NumPy arrays).
```

```python
#np.linspace(): Stands for Linearly Spaced. It generates an array of numbers
#  that are evenly distributed over a specified range. Unlike np.arange(), which 
# requires you to specify the step size (increment),
#  linspace requires you to specify the exact total number of values you want.
linspace_array = np.linspace(0, 1, 5)
print(linspace_array)
```
