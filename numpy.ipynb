# Array Indexing

## What does it do?
## Allows us to access individual elements.

```python
marks = np.array([23,45,12,34,12,34,56])
print(marks[0])
print(marks[6])
```

```python
data = np.array([
    [45,34,56],
    [34,56,23],
    [12,33,77]
])
print(data[0][1])
```

# Slicing
## What does it do?
## Selects a range of elements.


```python
marks = np.array([12,345,53,67])
print(marks[0:2])
```

## Selecting a column from a 2D array
### This means: 
### :  → all rows 
### 1  → column 1


```python
data = np.array([
    [45,34,56],
    [34,56,23],
    [12,33,77]
])
print(data[:,2])
```

## Arithmetic Operations
#### What does it do? 
#### Performs mathematical operations on all elements of an array.
#### This is called vectorized operation. 
#### Instead of writing a loop manually, NumPy performs the operation on the complete array.

```python
marks = np.array([12,234,46,45,67])
print(marks*2)
print(marks+2)
print(marks//2)
```

#### np.sum() 
##### What does it do? 
##### Calculates the total.
#### Use 
##### Useful for totals such as: 
##### 1.Total sales 
##### 2.Total revenue 
##### 3.Total quantities 
##### 4.Total scores

```python
marks = np.array([12,34,45,23,56])
print(np.sum(marks))
```

#### np.mean() 
##### What does it do? 
##### Calculates the average.

```python
marks = np.array([34,25,67,57,78])
print(np.mean(marks))
```

#### np.median() 
##### What does it do?
##### Finds the middle value after sorting the data.
##### Why is it useful? 
##### Median is often useful when data contains outliers.

```python
mark = np.array([34,23,45,67,12])
print(np.median(mark))
```

```python
marks = np.array([45,67,23,89,34])
print(np.min(marks))
print(np.max(marks))
```

##### np.std() 
###### What does it do? 
###### Calculates standard deviation.
###### Why is it useful? 
###### Standard deviation tells us how spread out values are around the mean. 
###### This becomes important later when learning: 
###### Data distribution 
###### Scaling 
###### Standardization
###### Variance 
###### Model preprocessing

```python
marks = np.array([23,67,23,79,90])
print(np.std(marks))
```

##### np.arange() 
##### What does it do? 
##### Creates evenly spaced values using a step.

```python
makrs = np.arange(1,15)
print(makrs)
```

```python
marks = np.arange(0,20,2) #skipps one number and starts with 2
print(marks)
```

##### np.linspace() 
###### What does it do? 
###### Creates a specified number of evenly spaced values between two limits.
###### Useful when we need evenly distributed numerical points, especially for mathematical or visualization tasks.

```python
a = np.linspace(1,4,5)
print(a)
```

#### np.reshape() 
##### What does it do? 
##### Changes the shape of an array without changing its data.
##### This is important because Machine Learning data is frequently represented as a matrix:

```python
a = np.array([23,46,67,24,67,89])
b = a.reshape(2,3)
print(b)
```

#### Boolean Filtering 
##### What does it do? 
##### Selects values based on a condition.
##### This introduces an important Data Science concept: Select data based on a condition.

```python
a = np.array([23,56,123,67,9,823])
print(a[a >= 0])
```
